```python
import pandas as pd

df = pd.read_csv("../MetObjects.csv", dtype=str)
```

```python
# Prints first five objects 
df.head()
```

```python
# Prints the column headers
df.columns
```

```python
# Filters to only show Object ID, Title and Object Name
df[["Object ID", "Title", "Object Name"]]
```

```python
# Shows index location 0
df.iloc[0]
```

```python
# Shows only objects with an Object Name that is Print
df[df["Object Name"] == 'Print']
```

```python
# Looks for objects from ONLY Denmark
df[df["Country"].isin(["Denmark"])]
```

```python
# Filters out object from United States
df[~df["Country"].isin(["United States"])]
```

```python
# Drops all rows containing missing values
df.dropna()
```

```python
# fills all missing data fields with 0
# you can also do df.fillna({"Object Name": '0'}) for it to only be done to Object Name column
# you can add inplace=True to modify existing dataframe instead of returning a new one
df.fillna('0')
```
